```python
import os
import json
```

```python
base_dir = "config"
tasks = [f"comparison_baseline_{i}" for i in ['tt', 'cv']]
model = "lstm"
model_configs = ["v1", "v2", "v3", "v4", "v5"]
interval_seconds_list = [5, 15, 30]
embedding_size = [32, 64, 128, 256, 320]
hidden_size = [32, 64, 128, 256, 320]
num_layers = 1
input_size_list = [100 * i for i in interval_seconds_list]
for task in tasks:
    for i, model_config in enumerate(model_configs):
        for interval, input_size in zip(interval_seconds_list, input_size_list):
            config_dict = {}
            config_dict['input_size'] = input_size
            config_dict['embedding_size'] = embedding_size[i]
            config_dict['hidden_size'] = hidden_size[i]
            config_dict['num_layers'] = num_layers
            filename = f"./{task}/{model}_{model_config}_{interval}sec_config.json"
            with open(filename, "w") as file:
                json.dump(config_dict, file)
```

```python
base_dir = "config"
tasks = [f"comparison_baseline_{i}" for i in ['tt', 'cv']]
model = "xlstm"
model_configs = ["v1", "v2", "v3", "v4", "v5"]

# model_configs = ["smaller"]
context_time_mins = 5
interval_seconds_list = [5, 15, 30]
embedding_dim = [32, 64, 128, 256, 320]
input_size_list = [100 * i for i in interval_seconds_list]
context_length_list = [int(context_time_mins * 60 // i) for i in interval_seconds_list]
conv1d_params = 4
qkv_size = 4
num_heads = 2
num_repeats = 1
for task in tasks:
    for i, model_config in enumerate(model_configs):
        if model_config == "default" or model_config == 'smaller':
            ratio = "1:1"
        elif model_config == "mlstm":
            ratio = "1:0"
        elif model_config == "slstm":
            ratio = "0:1"
        for interval, input_size, context_len in zip(interval_seconds_list, input_size_list, context_length_list):
            num_mlstm = int(ratio[0])
            num_slstm = int(ratio[-1])
            num_cells = num_mlstm + num_slstm
            if num_slstm == 1:
                if num_cells == 1:
                    slstm_at = [0]
                else:
                    slstm_at = [1]
            else:
                slstm_at = []
            config_dict = {}
            config_dict['input_size'] = input_size
            config_dict["conv1d_kernel_size"] = conv1d_params
            config_dict["qkv_proj_blocksize"] = qkv_size
            config_dict["num_heads"] = num_heads
            config_dict['context_length'] = context_len
            config_dict['num_blocks'] = num_cells
            config_dict['embedding_dim'] = embedding_dim[i]
            config_dict['slstm_at'] = slstm_at
            config_dict['num_reps'] = num_repeats
            filename = f"./{task}/{model}_{model_config}_{interval}sec_config.json"
            with open(filename, "w") as file:
                json.dump(config_dict, file)
```
